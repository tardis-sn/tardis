# Single-shell resonance explorer

This notebook follows one packet on a straight path through one shell. Each lesson chooses the radial velocity law, the line target and the launch position, then plots where the projected velocity of the packet meets the target. All lengths and velocities use illustrative units.

<div align="center">

| Symbol | Variable | Meaning |
|:---:|:---:|:---:|
| $r_\text{in}$, $r_\text{out}$ | `inner_radius`, `outer_radius` | Shell edges |
| $m$, $q$ | `velocity_slope`, `velocity_intercept` | Radial law $v(r) = m r + q$ |
| $b$ | `impact_parameter` | Closest distance of the ray to the centre |
| $x_0$ | `launch_position` | Signed launch coordinate along the ray |
| $w$ | `line_target_velocity` | Projected velocity of the line, $w = c\,(1 - \nu_\text{line} / \nu_\text{packet})$ |
| $s_\text{stop}$ | `scatter_distance` | Flight distance to an assumed electron scatter |

</div>

The position along the ray is $x = x_0 + s$, where $s$ is the flight distance. The radius is $r = \sqrt{x^2 + b^2}$. The packet sees the projected velocity

$$v_\parallel(x) = m x + \frac{q x}{\sqrt{x^2 + b^2}},$$

and its projected gradient is

$$G = m + \frac{q b^2}{r^3}.$$

A resonance is a point where $v_\parallel(x) = w$. The plot shows where resonances can occur. It does not compute whether the packet interacts with the line.

## Reading the plot

`calculate_distance_line_nonhomologous` in `tardis/transport/geometry/calculate_distances.py` squares the resonance equation into a quartic and returns its four roots. The legend in each plot names the markers for these roots.

```python
%matplotlib inline

from math import sqrt

from tardis.visualization.widgets.resonance_explorer.analysis import analyze_case
from tardis.visualization.widgets.resonance_explorer.model import ExplorerInputs
from tardis.visualization.widgets.resonance_explorer.render import render_guided_case

# Common shell and oblique packet ray shared by every lesson
inner_radius, outer_radius = 1.0, 3.0
impact_parameter = 1.2
outer_edge_launch = -sqrt(outer_radius**2 - impact_parameter**2)
```

## 1. Homologous expansion

**Inputs:** $m = 1$ and $q = 0$. The radial velocities at the shell edges are 1 and 3.

**In the plot:** The projected velocity is $v_\parallel = x$ and the gradient is $G = 1$ everywhere. The packet meets the target once, at $x = 0.3$. When $q = 0$, `solve_resonance_quartic` returns this root twice, so two markers sit at the same point. Its other two return values are NaN and have no marker.

```python
velocity_slope = 1.0
velocity_intercept = 0.0
line_target_velocity = 0.3
launch_position = outer_edge_launch

flight = ExplorerInputs(
    r_in=inner_radius,
    r_out=outer_radius,
    v_in=velocity_slope * inner_radius + velocity_intercept,
    v_out=velocity_slope * outer_radius + velocity_intercept,
    b=impact_parameter,
    x_0=launch_position,
    w=line_target_velocity,
)
homology_analysis = analyze_case(flight)

print(f"Shell edge velocities: v(r_in)={flight.v_in:g}, v(r_out)={flight.v_out:g}")
render_guided_case(homology_analysis)
```

## 2. Linked nonhomology

This lesson changes only $q$ from lesson 1. The plot draws lesson 1 as a comparison curve.

**Change from lesson 1:** $q = -2.34375$. The radial velocities at the shell edges become $-1.34375$ and $0.65625$.

**In the plot:** $G = 0$ at $x = -0.9$ and $x = 0.9$, so the projected velocity changes direction twice. The packet meets the target three times, at $x = -1.5559$, $-0.3415$ and $2.3955$.

The fourth real root, at $x = 0.1018$ (grey ▲), solves the squared equation

$$(m x - w)^2 (x^2 + b^2) - q^2 x^2 = 0$$

but not the original equation

$$m x + \frac{q x}{\sqrt{x^2 + b^2}} = w,$$

because squaring removes the sign of $q x / \sqrt{x^2 + b^2}$.

```python
velocity_slope = 1.0
velocity_intercept = -2.34375
line_target_velocity = 0.3
launch_position = outer_edge_launch

flight = ExplorerInputs(
    r_in=inner_radius,
    r_out=outer_radius,
    v_in=velocity_slope * inner_radius + velocity_intercept,
    v_out=velocity_slope * outer_radius + velocity_intercept,
    b=impact_parameter,
    x_0=launch_position,
    w=line_target_velocity,
)
analysis = analyze_case(flight)

print(f"Shell edge velocities: v(r_in)={flight.v_in:g}, v(r_out)={flight.v_out:g}")
render_guided_case(analysis, homology_analysis)
```

## 7. Special cases in the solver

Each case below makes the solver take a separate branch, or ends the flight with an electron scatter.

**Change from lesson 2:** $b = 0$ and $x_0 = -3$. The packet moves inward along a radius.

**In the plot:** `solve_resonance_quartic` takes its $b = 0$ branch. The packet meets the target once, at $x = -2.0438$. The flight ends at the inner shell edge, $x = -1$, so the packet does not reach the root at $x = 2.6437$ (amber ✕). The two roots at $x = 0$, the centre, are outside the shell (grey ▲).

```python
velocity_slope = 1.0
velocity_intercept = -2.34375
line_target_velocity = 0.3
launch_position = -3

flight = ExplorerInputs(
    r_in=inner_radius,
    r_out=outer_radius,
    v_in=velocity_slope * inner_radius + velocity_intercept,
    v_out=velocity_slope * outer_radius + velocity_intercept,
    b=0,  # radial ray
    x_0=launch_position,
    w=line_target_velocity,
)
analysis = analyze_case(flight)

print(f"Shell edge velocities: v(r_in)={flight.v_in:g}, v(r_out)={flight.v_out:g}")
render_guided_case(analysis)
```

**Inputs:** $m = 0$ and $q = 0.8$, so the radial velocity is 0.8 at both shell edges.

**In the plot:** `calculate_distance_line_nonhomologous` solves this case directly and does not call `solve_resonance_quartic`. It returns one root, at $x = 0.4854$, where the packet meets the target.
